# Case Study - PDAC (DIA and MRM pancreatic ductal adenocarcinoma serum data)

This is case study using the experimental data from [Kim et al. (2024)](https://www.mcponline.org/article/S1535-9476(24)00114-2/fulltext).

In this case study the authors first performed a discovery experiment using DIA.
After this they performed a targeted experiment analyzing a 15 protein subset of 
the DIA experiment using MRM. Here we analyze each dataset in turn and compare 
the variability across acquisition types.

# DIA
## Process MSstats Protein-Level Data

```python
pdac_dia_input = pd.read_csv('../data/pdac/PDAC_DIA_protein_level_data.csv')
# Remove proteins which could not be uniquely identified
pdac_dia_input = pdac_dia_input.loc[:, ~pdac_dia_input.columns.str.contains(';', na=False)]
```

### Convert protein names to gene IDs

```python
keep = ["originalRUN", "GROUP"]

# Convert UniProt to HGNC
orig_cols = list(pdac_dia_input.columns)
new_cols = []
dropped = []

for col in orig_cols:
    if col in keep:
        new_cols.append(col)
    else:
        try:
            new_name = hgnc_converter(col, mnemonic_id=True)
            new_cols.append(new_name)
        except Exception:
            dropped.append(col)

if dropped:
    pdac_dia_input = pdac_dia_input.drop(columns=dropped)

pdac_dia_input.columns = new_cols
pdac_dia_input = pdac_dia_input.drop(columns=[None])
pdac_dia_input.head()
```

### Remove proteins which were missing in majority of runs

```python
# Count missing values per column
missing_counts = pdac_dia_input.isna().sum()
missing_percentages = (missing_counts / len(pdac_dia_input))

print(f"Columns with missing values: {(missing_counts > 0).sum()}")
print(f"Total missing values: {missing_counts.sum()}")

# Remove columns with more than 50% missing values
threshold = 0.5
cols_to_keep = missing_percentages[missing_percentages <= threshold].index
pdac_dia_input = pdac_dia_input[cols_to_keep]

print(f"\nColumns after filtering: {len(pdac_dia_input.columns)}")
print(f"Columns removed: {len(missing_percentages) - len(cols_to_keep)}")
```

### Find missingness pattens

```python
# skip non-numeric columns
numeric = pdac_dia_input.select_dtypes(include=[np.number])
lod = numeric.quantile(.02, axis=0)

# Fit the MNAR detection curve on the real, pre-imputation data. originalRUN is a run
# label rather than an abundance, and fit_mnar casts everything else to float.
mnar_fits, mnar_params = fit_mnar(
    pdac_dia_input.drop(columns=['originalRUN']), group_col='GROUP', n_outer=60)
```

### Split train/validation

```python
X = pdac_dia_input.drop(columns=['GROUP'])
y = pdac_dia_input['GROUP']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=20, stratify=y
)

pdac_dia_input = pd.concat([y_train, X_train], axis=1)
validation_pdac_dia_input = pd.concat([y_test, X_test], axis=1)

print(f"Train: {pdac_dia_input.shape}, Test: {validation_pdac_dia_input.shape}")
print("Train class distribution:\n", pdac_dia_input['GROUP'].value_counts())
print("Test class distribution:\n", validation_pdac_dia_input['GROUP'].value_counts())
```

### Impute missing data using KNN

```python
# Keep original column order
orig_cols = pdac_dia_input.columns.tolist()

# Split numeric and non-numeric columns
numeric_cols = [c for c in orig_cols if pd.api.types.is_numeric_dtype(pdac_dia_input[c])]
non_numeric_cols = [c for c in orig_cols if c not in numeric_cols]

# Report missing values before imputation
print("Missing values before imputation (numeric cols):")
print(pdac_dia_input[numeric_cols].isna().sum().sum())

# KNN imputation
imputer = KNNImputer(n_neighbors=5, weights="uniform")
imputed_array = imputer.fit_transform(pdac_dia_input[numeric_cols])

# Build imputed DataFrame and restore original column order
imputed_numeric = pd.DataFrame(imputed_array, columns=numeric_cols, index=pdac_dia_input.index)
pdac_dia_input = pd.concat([pdac_dia_input[non_numeric_cols], imputed_numeric], axis=1)[orig_cols]

# Report missing values after imputation
print("\nMissing values after imputation (numeric cols):")
print(pdac_dia_input[numeric_cols].isna().sum().sum())
```

### Correlation Heatmaps

#### Experimental data

```python
def plot_corr(
    df: pd.DataFrame,
    size: float = 10,
    method: str = "pearson",
    cmap: str = "RdBu_r",
):
    """
    Professional, label-free correlation heatmap (lower triangle),
    optimized for large feature sets and colorblind safety.
    """
    # Keep numeric columns only
    x = df.select_dtypes(include=[np.number])
    if x.shape[1] < 2:
        raise ValueError("Need at least 2 numeric columns.")

    corr = x.corr(method=method)

    # Mask upper triangle + diagonal
    mask = np.triu(np.ones_like(corr, dtype=bool))
    corr_plot = corr.mask(mask)

    fig, ax = plt.subplots(figsize=(size, size), dpi=150)

    im = ax.imshow(
        corr_plot.to_numpy(),
        vmin=-1,
        vmax=1,
        cmap=cmap,
        interpolation="nearest",
    )

    # Remove all ticks and labels
    ax.set_xticks([])
    ax.set_yticks([])

    # Optional subtle grid for structure (can remove if you want ultra-clean)
    n = corr.shape[0]
    ax.set_xticks(np.arange(-0.5, n, 1), minor=True)
    ax.set_yticks(np.arange(-0.5, n, 1), minor=True)
    ax.grid(which="minor", linewidth=0.4, alpha=.2)
    ax.tick_params(which="minor", bottom=False, left=False)

    # Colorbar
    cbar = fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
    cbar.set_ticks([-1, -0.5, 0, 0.5, 1])
    cbar.ax.tick_params(labelsize=10)
    cbar.set_label("Correlation", fontsize=11)

    fig.tight_layout()
    return fig, ax
```

## $t$-test Power Calcuation

### Calculate experimental "positives"

```python
from MSquantSim.downstreamanalysis import run_ttest

# Ground-truth "positives": proteins called significant on the full experimental arms.
# assign() rather than in-place assignment so the simulation inputs above keep their
# original columns if these cells are re-run.
pdac_dia_experimental_all = pd.concat([
    pdac_dia_input_healthy.assign(GROUP="Normal"),
    pdac_dia_input_disease.assign(GROUP="Cancer"),
], axis=0)

da_results = run_ttest(
    pdac_dia_experimental_all, condition_col='GROUP', equal_var=False,
    adjust_pvalues=True, significance_level=0.05, min_samples=2,
    sort_by_pvalue=True)

positives = da_results.loc[da_results["significant"] == True]["Protein"].values
print(f"{len(positives)} significant proteins out of {len(da_results)} tested")
```

```python
copula_model_accuracy_fine = dict()
copula_rf_failed_fine = list()

for i in fine_n:
    temp_acc = list()
    for s in tqdm(range(100), desc=f"Running simulations: {i}"):
        training_data = copula_input_fine[i][s]
        
        X = training_data.drop("GROUP", axis=1)
        y = training_data["GROUP"]
        try:
            temp_acc.append(train_rf(X, y, validation_pdac_dia_input))
        except ValueError:
            # Unusable training data (e.g. DIA disease simulation 71 contains an inf value);
            # recorded as NaN, as the PCA cells do for the same iterations
            temp_acc.append(np.nan)
            copula_rf_failed_fine.append((i, s))
    copula_model_accuracy_fine[i] = temp_acc
copula_model_accuracy_fine = pd.DataFrame(copula_model_accuracy_fine)
copula_model_accuracy_fine.to_csv("../data/results/PDAC_DIA/copula_model_accuracy_fine.csv")
print("Failed (n, iteration):", copula_rf_failed_fine)
```

```python
def plot_corr(
    df: pd.DataFrame,
    size: float = 10,
    method: str = "pearson",
    cmap: str = "RdBu_r",
):
    """
    Professional, label-free correlation heatmap (lower triangle),
    optimized for large feature sets and colorblind safety.
    """
    # Keep numeric columns only
    x = df.select_dtypes(include=[np.number])
    if x.shape[1] < 2:
        raise ValueError("Need at least 2 numeric columns.")

    corr = x.corr(method=method)

    # Mask upper triangle + diagonal
    mask = np.triu(np.ones_like(corr, dtype=bool))
    corr_plot = corr.mask(mask)

    fig, ax = plt.subplots(figsize=(size, size), dpi=150)

    im = ax.imshow(
        corr_plot.to_numpy(),
        vmin=-1,
        vmax=1,
        cmap=cmap,
        interpolation="nearest",
    )

    # Remove all ticks and labels
    ax.set_xticks([])
    ax.set_yticks([])

    # Optional subtle grid for structure (can remove if you want ultra-clean)
    n = corr.shape[0]
    ax.set_xticks(np.arange(-0.5, n, 1), minor=True)
    ax.set_yticks(np.arange(-0.5, n, 1), minor=True)
    ax.grid(which="minor", linewidth=0.4, alpha=.2)
    ax.tick_params(which="minor", bottom=False, left=False)

    # Colorbar
    cbar = fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
    cbar.set_ticks([-1, -0.5, 0, 0.5, 1])
    cbar.ax.tick_params(labelsize=10)
    cbar.set_label("Correlation", fontsize=11)

    fig.tight_layout()
    return fig, ax
```

```python
# MRM validation set for the RF, prepared as in pdac_mrm_classification.py. There the >50%-missing
# filter and KNN imputation (k=5, uniform) are applied to the full MRM data *before* the stratified
# 80/20 split (random_state=20), and originalRUN is dropped. This gives the same validation rows and
# columns as validation_pdac_mrm_input above, with its missing values filled in.
# Used by the RF cells below and by the fine-grid RF; until 2026-10-09 the RF cells below
# passed validation_pdac_dia_input instead (bug I in plans/text_code_discrepancies.md).
pdac_mrm_rf = pd.read_csv('../data/pdac/PDAC_MRM_protein_level_data_v2.csv')

missing_percentages_rf = pdac_mrm_rf.isna().sum() / len(pdac_mrm_rf)
pdac_mrm_rf = pdac_mrm_rf[missing_percentages_rf[missing_percentages_rf <= 0.5].index]

orig_cols_rf = pdac_mrm_rf.columns.tolist()
numeric_cols_rf = [c for c in orig_cols_rf if pd.api.types.is_numeric_dtype(pdac_mrm_rf[c])]
non_numeric_cols_rf = [c for c in orig_cols_rf if c not in numeric_cols_rf]
imputed_rf = pd.DataFrame(
    KNNImputer(n_neighbors=5, weights="uniform").fit_transform(pdac_mrm_rf[numeric_cols_rf]),
    columns=numeric_cols_rf, index=pdac_mrm_rf.index)
pdac_mrm_rf = pd.concat([pdac_mrm_rf[non_numeric_cols_rf], imputed_rf], axis=1)[orig_cols_rf]

X_rf = pdac_mrm_rf.drop(columns=['GROUP', 'originalRUN'])
y_rf = pdac_mrm_rf['GROUP']
_, X_test_rf, _, y_test_rf = train_test_split(
    X_rf, y_rf, test_size=0.2, random_state=20, stratify=y_rf
)
validation_pdac_mrm_rf = pd.concat([y_test_rf, X_test_rf], axis=1)

# Same rows and columns as the notebook's (unimputed) MRM validation split, and the observed values agree
assert validation_pdac_mrm_rf.index.equals(validation_pdac_mrm_input.index)
assert list(validation_pdac_mrm_rf.columns) == list(validation_pdac_mrm_input.columns)
observed = validation_pdac_mrm_input.drop(columns='GROUP').notna()
assert np.allclose(validation_pdac_mrm_rf.drop(columns='GROUP')[observed].fillna(0),
                   validation_pdac_mrm_input.drop(columns='GROUP')[observed].fillna(0))

print(f"Validation: {validation_pdac_mrm_rf.shape}, "
      f"{int((~observed).sum().sum())} missing values imputed")
print(validation_pdac_mrm_rf['GROUP'].value_counts())
```

```python
import matplotlib.pyplot as plt

def publication_boxpanel(
    dfs,
    titles,
    *,
    xlabel="Sample size",
    ylabel="Accuracy",
    ylim=(0, 1),
    figsize=(20, 5.5),
    rotation=30,
    show_mean=True,
    savepath=None,   # e.g., "accuracy_panels.pdf" or "accuracy_panels.png"
    dpi=600,
):
    """
    Publication-style multi-panel boxplot figure.

    Parameters
    ----------
    dfs : list of pandas.DataFrame
        Each df should have columns corresponding to sample sizes (or categories).
        Values are accuracies.
    titles : list of str
        Panel titles, same length as dfs.
    savepath : str or None
        If given, figure is saved (PDF/SVG recommended). Otherwise just shown.
    """

    # --- Global typography / rcParams (minimal, journal-friendly) ---
    plt.rcParams.update({
        "font.size": 16,
        "axes.titlesize": 16,
        "axes.labelsize": 16,
        "xtick.labelsize": 16,
        "ytick.labelsize": 16,
        "axes.linewidth": 0.9,
        "xtick.major.width": 0.9,
        "ytick.major.width": 0.9,
        "xtick.major.size": 3.5,
        "ytick.major.size": 3.5,
        "figure.dpi": 120,
        "savefig.dpi": dpi,
        "savefig.bbox": "tight",
        "savefig.pad_inches": 0.02,
    })

    n = len(dfs)
    fig, axes = plt.subplots(
        1, n, figsize=figsize, sharey=True,
        gridspec_kw={"wspace": 0.12}
    )
    if n == 1:
        axes = [axes]

    for i, (df, title, ax) in enumerate(zip(dfs, titles, axes)):
        # Ensure deterministic column order
        cols = list(df.columns)

        ax.boxplot(
            [df[c].dropna().values for c in cols],
            labels=[str(c) for c in cols],
            patch_artist=True,
            showmeans=show_mean,
            meanprops=dict(marker="D", markersize=4, markerfacecolor="white",
                           markeredgewidth=0.9),
            medianprops=dict(linewidth=1.6),
            boxprops=dict(linewidth=0.9),
            whiskerprops=dict(linewidth=0.9),
            capprops=dict(linewidth=0.9),
            flierprops=dict(marker="o", markersize=2.3, markerfacecolor="none",
                            markeredgewidth=0.7, alpha=0.55),
        )

        # Style: white boxes + subtle outlines (prints well)
        for b in ax.artists:
            b.set_facecolor("white")
            b.set_edgecolor("black")

        ax.set_title(title, pad=8)
        ax.set_xlabel(xlabel)
        ax.set_ylim(*ylim)

        # Light y-grid only
        ax.grid(True, axis="y", linestyle="--", linewidth=0.7, alpha=0.35)
        ax.grid(False, axis="x")

        # Rotate tick labels gently (avoid 45 unless needed)
        ax.tick_params(axis="x", rotation=rotation)

        # Clean spines
        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)

        # Only left panel gets y-label
        if i == 0:
            ax.set_ylabel(ylabel)

    fig.align_ylabels(axes)

    if savepath:
        fig.savefig(savepath)
    plt.show()


# --- Usage (replace with your existing objects) ---
publication_boxpanel(
    dfs=[
        experimental_model_accuracy,
        copula_model_accuracy,
        tvae_model_accuracy,
        pprotein_model_accuracy,
    ],
    titles=[
        "Experimental",
        "Copula",
        "TVAE",
        "Per-protein",
    ],
    savepath=None,  # e.g. "accuracy_panels.pdf"
)
```

## $t$-test Power Calcuation

### Calculate experimental "positives"

```python
from MSquantSim.downstreamanalysis import run_ttest

# Ground-truth "positives": proteins called significant on the full experimental arms.
# assign() rather than in-place assignment so the simulation inputs above keep their
# original columns if these cells are re-run.
pdac_mrm_experimental_all = pd.concat([
    pdac_mrm_input_healthy.assign(GROUP="Normal"),
    pdac_mrm_input_disease.assign(GROUP="Cancer"),
], axis=0)

da_results = run_ttest(
    pdac_mrm_experimental_all, condition_col='GROUP', equal_var=False,
    adjust_pvalues=True, significance_level=0.05, min_samples=2,
    sort_by_pvalue=True)

positives = da_results.loc[da_results["significant"] == True]["Protein"].values
print(f"{len(positives)} significant proteins out of {len(da_results)} tested")
```

### Experimental

```python
experimental_positive_counts = dict()
experimental_valid_iterations = dict()

for i in n_samples:
    counts = pd.Series(0, index=positives)
    n_valid = 0
    for s in tqdm(range(100), desc=f"Running simulations: {i}"):
        training_data = experimental_input[i][s]
        try:
            da_results = run_ttest(training_data, condition_col='GROUP')
        except ValueError:
            # Subsample landed on a single condition, so no test is possible
            continue
        n_valid += 1
        sig_proteins = da_results.loc[da_results["significant"] == True, "Protein"]
        counts += counts.index.isin(sig_proteins)
    experimental_positive_counts[i] = counts
    experimental_valid_iterations[i] = n_valid

experimental_positive_counts = pd.DataFrame(experimental_positive_counts)
experimental_positive_counts.index.name = "Protein"
experimental_positive_counts = experimental_positive_counts.reset_index()

print("Iterations with both conditions present:", experimental_valid_iterations)
```

## Power curves: MRM vs DIA (semi-synthetic benchmark)

Plan 3.7 (`plans/acquisition_power.md`). Empirical $t$-test power and false discovery proportion (FDP) for PDAC-MRM and PDAC-DIA, from a semi-synthetic benchmark reference (the real data with the condition effect removed) and from simulators fitted to the same data (Copula, Per-protein). Each experiment draws 2N rows from a null pool, splits them into two groups, adds a known log2FC to k random proteins in one group, and calls proteins with Welch $t$-tests and BH adjustment at FDR 0.05 (`MSquantSim.power.run_experiments`, same test settings as `run_ttest`).

- Data: all samples (train + validation) of each acquisition, rebuilt here from the protein-level files with the same filtering as at the top of each part, KNN-imputed (k = 5) and centered within condition.
- Protein sets: MRM-15; DIA-15 (the same 15 proteins, matched by gene symbol); DIA-397 (secondary).
- Grid: log2FC ∈ {0.58, 1}; k = ceil(π₁ · p) for π₁ ∈ {0.05, 0.10, 0.20}; 500 experiments per cell.
- The section only needs the imports cell at the top of the notebook.

```python
from MSquantSim.power import (center_by_group, within_group_variance, run_experiments,
                              summarize_experiments)
from MSquantSim.simulate import fit_copula, sample_copula, fit_per_protein, sample_per_protein


def drop_mostly_missing(df, threshold=0.5):
    """Remove proteins missing in more than half of the runs (as in the filtering cells above)."""
    missing = df.isna().sum() / len(df)
    return df[missing[missing <= threshold].index]


def knn_impute(df, k=5):
    """KNN-impute the protein columns (k = 5, uniform weights, as in the imputation cells above)."""
    prots = [c for c in df.columns if pd.api.types.is_numeric_dtype(df[c])]
    out = df.copy()
    out[prots] = KNNImputer(n_neighbors=k, weights="uniform").fit_transform(df[prots])
    return out


# DIA, all samples: same steps as the top of the DIA part (drop ambiguous protein groups,
# UniProt mnemonic -> HGNC symbol, drop proteins missing in more than 50% of runs)
dia_all_raw = pd.read_csv('../data/pdac/PDAC_DIA_protein_level_data.csv')
dia_all_raw = dia_all_raw.loc[:, ~dia_all_raw.columns.str.contains(';', na=False)]
dia_new_cols, dia_dropped = [], []
for col in dia_all_raw.columns:
    if col in ["originalRUN", "GROUP"]:
        dia_new_cols.append(col)
    else:
        try:
            dia_new_cols.append(hgnc_converter(col, mnemonic_id=True))
        except Exception:
            dia_dropped.append(col)
dia_all_raw = dia_all_raw.drop(columns=dia_dropped)
dia_all_raw.columns = dia_new_cols
dia_all_raw = dia_all_raw.drop(columns=[None])
dia_all_raw = drop_mostly_missing(dia_all_raw).drop(columns=['originalRUN'])

# MRM, all samples
mrm_all_raw = drop_mostly_missing(
    pd.read_csv('../data/pdac/PDAC_MRM_protein_level_data_v2.csv')).drop(columns=['originalRUN'])

dia_all_imp = knn_impute(dia_all_raw)
mrm_all_imp = knn_impute(mrm_all_raw)

# The 15 MRM proteins, all of which are among the filtered DIA proteins
shared_proteins = [c for c in mrm_all_raw.columns if c != "GROUP"]
assert set(shared_proteins) <= set(dia_all_raw.columns)
assert not dia_all_raw.columns.duplicated().any()

for name, df in [("DIA", dia_all_raw), ("MRM", mrm_all_raw)]:
    prots = df.drop(columns="GROUP")
    print(f"{name}: {df.shape[0]} samples {df['GROUP'].value_counts().to_dict()}, "
          f"{prots.shape[1]} proteins, {prots.isna().mean().mean():.1%} missing "
          f"({prots[shared_proteins].isna().mean().mean():.1%} on the 15 shared proteins)")
```

### Step 0: within-condition SD, MRM vs DIA

```python
# Per-protein within-condition SD (pooled over Normal and Cancer) on the 15 shared proteins,
# all samples. Raw: missing values ignored; imputed: after KNN imputation.
step0_sd = pd.DataFrame({
    "sd_mrm_raw": np.sqrt(within_group_variance(mrm_all_raw[shared_proteins + ["GROUP"]], "GROUP")),
    "sd_dia_raw": np.sqrt(within_group_variance(dia_all_raw[shared_proteins + ["GROUP"]], "GROUP")),
    "sd_mrm_imputed": np.sqrt(within_group_variance(mrm_all_imp[shared_proteins + ["GROUP"]], "GROUP")),
    "sd_dia_imputed": np.sqrt(within_group_variance(dia_all_imp[shared_proteins + ["GROUP"]], "GROUP")),
})
# Paired differences: negative means MRM varies less than DIA
step0_sd["diff_raw"] = step0_sd["sd_mrm_raw"] - step0_sd["sd_dia_raw"]
step0_sd["diff_imputed"] = step0_sd["sd_mrm_imputed"] - step0_sd["sd_dia_imputed"]
step0_sd.index.name = "Protein"
step0_sd.to_csv("../data/results/PDAC_MRM/step0_sd_comparison.csv")

for kind in ["raw", "imputed"]:
    d = step0_sd[f"diff_{kind}"]
    print(f"{kind:8s} median SD: MRM {step0_sd[f'sd_mrm_{kind}'].median():.3f}, "
          f"DIA {step0_sd[f'sd_dia_{kind}'].median():.3f}; median paired difference "
          f"(MRM - DIA) {d.median():+.3f}; MRM lower for {(d < 0).sum()} of {len(d)} proteins")
step0_sd.round(3)
```

### Null pools and simulators

```python
# Null data: imputed data centered within condition (removes the condition effect)
power_null = {
    "MRM-15": center_by_group(mrm_all_imp[shared_proteins + ["GROUP"]], "GROUP"),
    "DIA-15": center_by_group(dia_all_imp[shared_proteins + ["GROUP"]], "GROUP"),
    "DIA-397": center_by_group(dia_all_imp, "GROUP"),
}
sim_pool_size = 20000
power_set_ids = {"MRM-15": 0, "DIA-15": 1, "DIA-397": 2}
power_method_ids = {"Semi-synthetic benchmark": 0, "Copula": 1, "Per-protein": 2}

# Semi-synthetic benchmark draws from the centered data itself. Each simulator is fit once per protein set
# on the same centered data, and experiments draw from a pool of 20,000 simulated rows.
power_pools = {}
for pset, null in power_null.items():
    seed = 100 + power_set_ids[pset]
    power_pools[(pset, "Semi-synthetic benchmark")] = null.to_numpy()
    power_pools[(pset, "Copula")] = sample_copula(
        fit_copula(null), sim_pool_size, random_state=seed)[null.columns].to_numpy(dtype=float)
    power_pools[(pset, "Per-protein")] = sample_per_protein(
        fit_per_protein(null), sim_pool_size,
        rng=np.random.default_rng(seed))[null.columns].to_numpy(dtype=float)

for (pset, method), pool in power_pools.items():
    finite = np.isfinite(pool)
    print(f"{pset:8s} {method:12s} pool {pool.shape}, non-finite values: {(~finite).sum()}, "
          f"median SD {np.median(np.nanstd(np.where(finite, pool, np.nan), axis=0, ddof=1)):.3f}, "
          f"max |value| {np.abs(pool[finite]).max():.2f}")
```

### Experiments

```python
power_methods = ["Semi-synthetic benchmark", "Copula", "Per-protein"]
power_log2fc = [0.58, 1.0]
power_pi1 = [0.05, 0.10, 0.20]
# N per group: MRM and DIA-15 matched comparison {3, ..., 30}; MRM additionally up to 100
power_n = {
    "MRM-15": [3, 5, 10, 15, 20, 25, 30, 40, 50, 75, 100],
    "DIA-15": [3, 5, 10, 15, 20, 25, 30],
    "DIA-397": [3, 5, 10, 15, 20, 25, 30],
}
n_power_experiments = 500
power_setting_cols = ["protein_set", "method", "log2fc", "pi1", "n_changed", "n_per_group"]


def run_power_grid(pools, n_grid, n_experiments):
    """All experiments for every (protein set, method, log2FC, pi1, N) in the grid."""
    results = []
    for pset, ns in n_grid.items():
        for method in power_methods:
            pool = pools[(pset, method)]
            for fc_i, fc in enumerate(power_log2fc):
                for pi_i, pi1 in enumerate(power_pi1):
                    # k = ceil(pi1 * p); round() keeps 0.2 * 15 = 3.0000000000000004 at 3
                    k = int(np.ceil(round(pi1 * pool.shape[1], 6)))
                    for n in ns:
                        # One seeded generator per (protein set, method, setting, N)
                        rng = np.random.default_rng(
                            [power_set_ids[pset], power_method_ids[method], fc_i, pi_i, n])
                        exp, _ = run_experiments(pool, n, k, fc, n_experiments, rng)
                        exp.insert(0, "experiment", np.arange(n_experiments))
                        settings = dict(zip(power_setting_cols, [pset, method, fc, pi1, k, n]))
                        results.append(exp.assign(**settings))
    columns = power_setting_cols + ["experiment", "tp", "fp", "discoveries", "power", "fdp"]
    return pd.concat(results, ignore_index=True)[columns]


def summarize_power_grid(experiments):
    """summarize_experiments for every (protein set, method, log2FC, pi1, N) cell."""
    return (experiments.groupby(power_setting_cols, sort=False)[["power", "fdp"]]
            .apply(summarize_experiments).reset_index())
```

```python
# Smoke test (optional, seconds): one N per protein set, 20 experiments. Nothing is saved.
power_smoke = run_power_grid(power_pools, {pset: [10] for pset in power_n}, 20)
print(power_smoke.shape)
summarize_power_grid(power_smoke).round(3)
```

```python
# Full run: 500 experiments per (protein set, method, log2FC, pi1, N)
power_curves_experiments = run_power_grid(power_pools, power_n, n_power_experiments)
power_curves_summary = summarize_power_grid(power_curves_experiments)

power_curves_experiments.to_csv("../data/results/PDAC_MRM/power_curves_experiments.csv", index=False)
power_curves_summary.to_csv("../data/results/PDAC_MRM/power_curves_summary.csv", index=False)
print(power_curves_experiments.shape, power_curves_summary.shape)
```

## Placement benchmark (4.2)

Plan `plans/placement_benchmark.md` (approved by Devon 2026-10-19). Does the placement of the differentially abundant (DE) proteins, random vs correlated modules, change the distributions of power and FDP when the number of DE proteins K is held fixed? Fully semi-synthetic: no Copula sampling.

- Data: the 4.2 data of the seed section (cell 260, `seed_data`): PDAC-DIA, 79 samples (40 Cancer, 39 Normal; run `1E446_Normal` excluded for 4.2), 397 proteins, KNN-imputed (k = 5), each protein centered within condition. Pool = this null (79 × 397).
- One experiment: 2n pool rows, split into two groups of n; the truth is added to group 2; Welch $t$-tests + BH at FDR 0.05 (`MSquantSim.power.run_experiments`).
- Noise correlation: **correlated** = the real null rows; **uncorrelated** = each protein draws its own 2n of its 79 values (`independent_columns=True`), which keeps every protein's values and removes the correlation between proteins.
- Truth (K fixed): **clustered** (`clustered_de_sets`, Devon's rule: a seed drawn uniformly from the unselected proteins plus its unselected partners with copula r > 0.5, at most the 15 strongest, repeated until K are selected; the last cluster trimmed to exactly K by dropping its weakest partners; one random sign per cluster; a partnerless seed is a singleton) or **random** (`random_de_set`: K uniform proteins, a random sign each; logged as K clusters of size 1). |log2FC| added with the truth's signs.
- Correlation matrix for the clusters: the fitted Gaussian copula correlation of the null (`fit_copula(null).correlation`, reindexed to the null's protein order).
- Grid: placement {random, clustered} × correlation {correlated, uncorrelated} × K {20, 40, 79} (5, 10, 20% of 397) × log2FC {0.58, 1} × n per group {3, 5, 8, 10, 15, 20, 25, 30, 39} = 216 cells, R = 500 replicates each. Above n = 20 the correlated experiments reuse most of the 79 samples, which understates the spread between replicates.
- Seeds (`np.random.SeedSequence(20261019, spawn_key=...)`, indices into the grids): truths `spawn_key=(0, placement, K, n)`, drawn once per (placement, K, n), a new truth per replicate, shared by both correlation levels and both log2FC (paired); rows `spawn_key=(1, placement, correlation, K, log2FC, n)`, one stream per cell.
- Logged per replicate: number of clusters, singleton clusters and their fraction, mean within-group variance of the DE set (`within_group_variance(seed_data, "GROUP")`), mean |copula r| over the DE-set pairs.
- Standard calculator: `analytic_sample_size` with the median within-group variance of `seed_data`, π₁ = K / 397, FDR 0.05, power 0.8.
- Compute: joblib over grid cells with 4 workers and 1 BLAS thread each; the Copula fit runs with at most 4 BLAS threads.
- Outputs: `../data/results/PDAC_DIA/placement_{replicates,clusters,summary,calculator}.csv`.
- To reproduce: `run_cells.py PDAC_case_study.ipynb 2,247,260,258,261-267` (cell numbers as of 2026-10-19).

### Truth assignments and experiments

Truths are drawn in the main process (one stream per (placement, K, n)); the experiments run in parallel, one task per grid cell, each with its own row stream. The smoke test uses the first 20 truths and rows of the same streams.

```python
# Grid, truth assignments and the parallel experiment loop
pl_entropy = 20261019
pl_fdr = 0.05
pl_placements = ["random", "clustered"]
pl_correlations = ["correlated", "uncorrelated"]
pl_K_grid = [20, 40, 79]
pl_log2fc_grid = [0.58, 1.0]
pl_n_grid = [3, 5, 8, 10, 15, 20, 25, 30, 39]   # 2n <= 79 pool rows
pl_R = 500
pl_cell_cols = ["placement", "correlation", "K", "log2fc", "n_per_group"]
pl_abs_corr = np.abs(pl_corr)


def pl_draw_truths(placement, K, n, n_reps):
    """n_reps truth assignments for one (placement, K, n), with their per-replicate logs."""
    key = (0, pl_placements.index(placement), pl_K_grid.index(K), pl_n_grid.index(n))
    rng = np.random.default_rng(np.random.SeedSequence(pl_entropy, spawn_key=key))
    sets, signs, logs, clusters = [], [], [], []
    for r in range(n_reps):
        if placement == "clustered":
            idx, sgn, sizes = clustered_de_sets(pl_corr, K, rng, min_corr=0.5, max_partners=15)
            clusters += [dict(K=K, n_per_group=n, replicate=r, cluster_index=c, size=s)
                         for c, s in enumerate(sizes)]
        else:
            idx, sgn = random_de_set(pl_p, K, rng)
            sizes = [1] * K
        sub = pl_abs_corr[np.ix_(idx, idx)]
        n_single = sum(s == 1 for s in sizes)
        sets.append(idx)
        signs.append(sgn)
        logs.append(dict(replicate=r, n_clusters=len(sizes), n_singletons=n_single,
                         frac_singletons=n_single / len(sizes),
                         de_mean_variance=pl_variance[idx].mean(),
                         de_mean_abs_r=sub[np.triu_indices(K, 1)].mean()))
    return sets, signs, pd.DataFrame(logs), pd.DataFrame(clusters)


def pl_run_cell(pool, placement, correlation, K, log2fc, n, sets, signs, key):
    """All replicates of one grid cell (runs in a joblib worker, 1 BLAS thread)."""
    with threadpool_limits(1):
        rng = np.random.default_rng(np.random.SeedSequence(pl_entropy, spawn_key=key))
        exps, _ = run_experiments(pool, n, None, log2fc, len(sets), rng, fdr=pl_fdr,
                                  changed_sets=sets, changed_signs=signs,
                                  independent_columns=(correlation == "uncorrelated"))
    exps.insert(0, "replicate", np.arange(len(sets)))
    return exps.assign(placement=placement, correlation=correlation, K=K, log2fc=log2fc,
                       n_per_group=n)


def pl_run_grid(K_grid, n_grid, n_reps, n_jobs=4):
    """Replicates, cluster sizes and the per-cell summary for every cell of the grid."""
    truths, logs, clusters, tasks = {}, [], [], []
    for placement in pl_placements:
        for K in K_grid:
            for n in n_grid:
                sets, signs, log, clu = pl_draw_truths(placement, K, n, n_reps)
                truths[(placement, K, n)] = (sets, signs)
                logs.append(log.assign(placement=placement, K=K, n_per_group=n))
                clusters.append(clu)
    for placement in pl_placements:
        for correlation in pl_correlations:
            for K in K_grid:
                for log2fc in pl_log2fc_grid:
                    for n in n_grid:
                        key = (1, pl_placements.index(placement), pl_correlations.index(correlation),
                               pl_K_grid.index(K), pl_log2fc_grid.index(log2fc), pl_n_grid.index(n))
                        tasks.append((placement, correlation, K, log2fc, n,
                                      *truths[(placement, K, n)], key))
    results = Parallel(n_jobs=n_jobs)(delayed(pl_run_cell)(pl_pool, *t) for t in tasks)
    reps = (pd.concat(results, ignore_index=True)
            .merge(pd.concat(logs, ignore_index=True), on=["placement", "K", "n_per_group", "replicate"],
                   how="left", validate="many_to_one")
            .rename(columns={"power": "tpr"}))
    reps = reps[pl_cell_cols + ["replicate", "discoveries", "tp", "fp", "fdp", "tpr", "n_clusters",
                                "n_singletons", "frac_singletons", "de_mean_variance", "de_mean_abs_r"]]
    summary = (reps.groupby(pl_cell_cols, sort=False)
               .apply(lambda g: pd.concat([
                   summarize_success(g.rename(columns={"tpr": "power"}), fdr=pl_fdr, target_power=0.8),
                   pd.Series({"mean_n_clusters": g["n_clusters"].mean(),
                              "mean_frac_singletons": g["frac_singletons"].mean(),
                              "mean_de_variance": g["de_mean_variance"].mean(),
                              "mean_de_abs_r": g["de_mean_abs_r"].mean()})]),
                      include_groups=False)
               .reset_index())
    return reps, pd.concat(clusters, ignore_index=True), summary
```

### Ground truth vs Copula vs standard calculator

Addition to the placement benchmark (`plans/placement_benchmark.md`, "Addition: ground truth vs Copula vs standard calculator"; approved by Devon 2026-10-19). The ground truth is the real-noise benchmark above (`placement_replicates.csv`, correlation = "correlated"; not re-run). Two methods get exactly the ground truth's truth assignments:

- **copula:** 2n rows from 20,000 rows sampled from the Copula fitted in cell 261 (`pl_copula`, no refit; `sample_copula(..., random_state=20261019)` with at most 4 BLAS threads).
- **calculator_sim:** the standard calculator's assumptions: 20,000 rows of independent Normal(0, σ²) for all 397 proteins, σ² = the median within-group variance of `seed_data` (as in `placement_calculator.csv`); rng `SeedSequence(20261019, spawn_key=(3,))`.
- Truths: regenerated with the scheme of cell 263 (spawn key `(0, placement, K, n)`); for the ground-truth n the DE sets and signs are checked against `pl_draw_truths` and the per-replicate summaries against `placement_replicates.csv`. n ∈ {50, 75, 100} continue the n index (9, 10, 11) and have no ground truth.
- Grid: method {copula, calculator_sim} × placement {random, clustered} × K {20, 40, 79} × log2FC {1, 0.58} × n {3, 5, 8, 10, 15, 20, 25, 30, 39, 50, 75, 100}, R = 500; rows from spawn key `(2, method, placement, K, log2FC, n)`; real-row draws (`independent_columns=False`); Welch $t$-tests + BH at 0.05; 4 joblib workers, 1 BLAS thread each.
- Outputs: `../data/results/PDAC_DIA/placement_methods_{replicates,summary}.csv`.
- To reproduce: `run_cells.py PDAC_case_study.ipynb 2,247,260-261,263,268-274` (cell numbers as of 2026-10-21, after the reorder).

```python
# Pools: copula (20,000 rows from the Copula fitted in cell 261) and calculator_sim (independent
# Normal(0, median within-group variance) for all 397 proteins)
from MSquantSim.simulate import sample_copula

pm_pool_size = 20000
pm_median_variance = float(np.median(pl_variance))
t0 = time.time()
with threadpool_limits(4):
    pm_copula_pool = sample_copula(pl_copula, pm_pool_size,
                                   random_state=20261019)[pl_prots].to_numpy(dtype=float)
pm_sampling_time = time.time() - t0
pm_calc_rng = np.random.default_rng(np.random.SeedSequence(pl_entropy, spawn_key=(3,)))
pm_pools = {"copula": pm_copula_pool,
            "calculator_sim": pm_calc_rng.normal(0.0, np.sqrt(pm_median_variance),
                                                 size=(pm_pool_size, pl_p))}
print(f"Copula sampling (20,000 rows): {pm_sampling_time:.0f} s; calculator_sim variance {pm_median_variance:.4f}")

pm_real_max = np.abs(pl_pool).max()
pm_pool_check = []
for name, values in {"ground_truth (real null)": pl_pool, **pm_pools}.items():
    finite = np.isfinite(values)
    check = dict(pool=name, rows=values.shape[0], non_finite=int((~finite).sum()),
                 median_sd=np.median(np.nanstd(np.where(finite, values, np.nan), axis=0, ddof=1)),
                 max_abs_value=np.abs(values[finite]).max(),
                 n_beyond_real_max=int((np.abs(values[finite]) > pm_real_max).sum()))
    if finite.all():
        check["mean_offdiag_r"] = np.corrcoef(values, rowvar=False)[pl_iu].mean()
    pm_pool_check.append(check)
pm_pool_check = pd.DataFrame(pm_pool_check).set_index("pool")
display(pm_pool_check.round(4))
assert pm_pool_check.loc["copula", "non_finite"] == 0, "non-finite values in the copula pool: stop and report"
```

```python
# Truths for every n (same scheme as pl_draw_truths; the extra n continue the n index), and the
# check that the ground-truth n reproduce the saved ground truth exactly
pm_n_grid = pl_n_grid + [50, 75, 100]   # 50-100: no ground truth (79-sample pool)
pm_methods = ["copula", "calculator_sim"]


def pm_truths(placement, K, n, n_reps):
    """Truths and per-replicate logs for one (placement, K, n); spawn key (0, placement, K, n)."""
    key = (0, pl_placements.index(placement), pl_K_grid.index(K), pm_n_grid.index(n))
    rng = np.random.default_rng(np.random.SeedSequence(pl_entropy, spawn_key=key))
    sets, signs, logs = [], [], []
    for r in range(n_reps):
        if placement == "clustered":
            idx, sgn, sizes = clustered_de_sets(pl_corr, K, rng, min_corr=0.5, max_partners=15)
        else:
            idx, sgn = random_de_set(pl_p, K, rng)
            sizes = [1] * K
        sub = pl_abs_corr[np.ix_(idx, idx)]
        n_single = sum(s == 1 for s in sizes)
        sets.append(idx)
        signs.append(sgn)
        logs.append(dict(replicate=r, n_clusters=len(sizes), n_singletons=n_single,
                         frac_singletons=n_single / len(sizes),
                         de_mean_variance=pl_variance[idx].mean(),
                         de_mean_abs_r=sub[np.triu_indices(K, 1)].mean()))
    return sets, signs, pd.DataFrame(logs)


pm_saved = pd.read_csv("../data/results/PDAC_DIA/placement_replicates.csv",
                       float_precision="round_trip").query("correlation == 'correlated'")
pm_checks = []
for placement in pl_placements:
    for K in pl_K_grid:
        for n in pl_n_grid:
            sets, signs, log = pm_truths(placement, K, n, pl_R)
            ref_sets, ref_signs, _, _ = pl_draw_truths(placement, K, n, pl_R)
            m = (pm_saved.query("placement == @placement and K == @K and n_per_group == @n")
                 .merge(log, on="replicate", suffixes=("_saved", "")))
            pm_checks.append(dict(
                placement=placement, K=K, n_per_group=n, saved_rows=len(m),
                sets_signs_equal=all(np.array_equal(a, b) for a, b in zip(sets, ref_sets))
                                 and all(np.array_equal(a, b) for a, b in zip(signs, ref_signs)),
                n_clusters_equal=bool((m["n_clusters_saved"] == m["n_clusters"]).all()),
                de_mean_variance_equal=bool((m["de_mean_variance_saved"] == m["de_mean_variance"]).all()),
                de_mean_abs_r_equal=bool((m["de_mean_abs_r_saved"] == m["de_mean_abs_r"]).all()),
                max_abs_diff=float(max((m["de_mean_variance_saved"] - m["de_mean_variance"]).abs().max(),
                                       (m["de_mean_abs_r_saved"] - m["de_mean_abs_r"]).abs().max()))))
pm_truth_check = pd.DataFrame(pm_checks)
pm_truth_ok = bool(pm_truth_check[["sets_signs_equal", "n_clusters_equal", "de_mean_variance_equal",
                                   "de_mean_abs_r_equal"]].all().all()
                   and (pm_truth_check["saved_rows"] == 2 * pl_R).all())
print(f"truth check over {len(pm_truth_check)} (placement, K, n) combinations x {pl_R} replicates "
      f"(saved correlated rows of both log2FC): all equal = {pm_truth_ok}; "
      f"max |difference| {pm_truth_check['max_abs_diff'].max():.3g}")
assert pm_truth_ok, "truths do not reproduce the saved ground truth: stop and report"
```

```python
# Experiments on the copula and calculator_sim pools with the ground truth's truths; one joblib
# task per cell (pl_run_cell from cell 263, real-row draws)
pm_cell_cols = ["method", "placement", "K", "log2fc", "n_per_group"]


def pm_run_grid(K_grid, n_grid, n_reps, n_jobs=4):
    """Replicates and the per-cell summary for every (method, placement, K, log2FC, n)."""
    truths, logs, tasks = {}, [], []
    for placement in pl_placements:
        for K in K_grid:
            for n in n_grid:
                sets, signs, log = pm_truths(placement, K, n, n_reps)
                truths[(placement, K, n)] = (sets, signs)
                logs.append(log.assign(placement=placement, K=K, n_per_group=n))
    for method in pm_methods:
        for placement in pl_placements:
            for K in K_grid:
                for log2fc in pl_log2fc_grid:
                    for n in n_grid:
                        key = (2, pm_methods.index(method), pl_placements.index(placement),
                               pl_K_grid.index(K), pl_log2fc_grid.index(log2fc), pm_n_grid.index(n))
                        tasks.append((method, placement, K, log2fc, n, *truths[(placement, K, n)], key))
    results = Parallel(n_jobs=n_jobs)(
        delayed(pl_run_cell)(pm_pools[method], placement, "correlated", K, log2fc, n, sets, signs, key)
        for method, placement, K, log2fc, n, sets, signs, key in tasks)
    reps = (pd.concat([res.drop(columns="correlation").assign(method=task[0])
                       for res, task in zip(results, tasks)], ignore_index=True)
            .merge(pd.concat(logs, ignore_index=True), on=["placement", "K", "n_per_group", "replicate"],
                   how="left", validate="many_to_one")
            .rename(columns={"power": "tpr"}))
    reps = reps[pm_cell_cols + ["replicate", "discoveries", "tp", "fp", "fdp", "tpr", "n_clusters",
                                "n_singletons", "frac_singletons", "de_mean_variance", "de_mean_abs_r"]]
    summary = (reps.groupby(pm_cell_cols, sort=False)
               .apply(lambda g: summarize_success(g.rename(columns={"tpr": "power"}), fdr=pl_fdr,
                                                  target_power=0.8), include_groups=False)
               .reset_index())
    return reps, summary
```

```python
# Experiments on the perprotein_sim pool with the ground truth's truths (pm_truths, cell 270);
# one joblib task per cell (pl_run_cell, cell 263)
pp_method, pp_method_id = "perprotein_sim", 2
assert pp_method not in pm_methods and pp_method_id >= len(pm_methods)   # copula = 0, calculator_sim = 1
pp_cell_cols = ["method", "placement", "K", "log2fc", "n_per_group"]
pp_truth_cols = ["n_clusters", "n_singletons", "frac_singletons", "de_mean_variance", "de_mean_abs_r"]


def pp_run_grid(K_grid, n_grid, n_reps, n_jobs=4):
    """Replicates and the per-cell summary of perprotein_sim for every (placement, K, log2FC, n)."""
    truths, logs, tasks = {}, [], []
    for placement in pl_placements:
        for K in K_grid:
            for n in n_grid:
                sets, signs, log = pm_truths(placement, K, n, n_reps)
                truths[(placement, K, n)] = (sets, signs)
                logs.append(log.assign(placement=placement, K=K, n_per_group=n))
    for placement in pl_placements:
        for K in K_grid:
            for log2fc in pl_log2fc_grid:
                for n in n_grid:
                    key = (2, pp_method_id, pl_placements.index(placement), pl_K_grid.index(K),
                           pl_log2fc_grid.index(log2fc), pm_n_grid.index(n))
                    tasks.append((placement, K, log2fc, n, *truths[(placement, K, n)], key))
    results = Parallel(n_jobs=n_jobs)(
        delayed(pl_run_cell)(pp_pool, placement, "correlated", K, log2fc, n, sets, signs, key)
        for placement, K, log2fc, n, sets, signs, key in tasks)
    reps = (pd.concat([res.drop(columns="correlation").assign(method=pp_method) for res in results],
                      ignore_index=True)
            .merge(pd.concat(logs, ignore_index=True), on=["placement", "K", "n_per_group", "replicate"],
                   how="left", validate="many_to_one")
            .rename(columns={"power": "tpr"}))
    reps = reps[pp_cell_cols + ["replicate", "discoveries", "tp", "fp", "fdp", "tpr"] + pp_truth_cols]
    summary = (reps.groupby(pp_cell_cols, sort=False)
               .apply(lambda g: summarize_success(g.rename(columns={"tpr": "power"}), fdr=pl_fdr,
                                                  target_power=0.8), include_groups=False)
               .reset_index())
    return reps, summary
```

### Step 0: premise check with the real labels

For the 80-sample data (`dia_all_imp`, as in 3.7) and the 79-sample 4.2 data: D = proteins differentially abundant between Cancer and Normal (Welch $t$-test on the imputed data, BH at FDR 0.05). For two within-group correlation matrices, the centered data (all samples) and the Normal samples only (disease heterogeneity within the Cancer group can itself create correlation):
- share of D among the top-3 positive partners of D proteins vs of non-D proteins, vs the base rate |D| / 397;
- P(protein ∈ D | it has r ≥ 0.5 with some other D protein), vs the base rate;
- permutation baseline: 1,000 random sets of size |D|;
- correlation noise: off-diagonal r, per-protein maximum r, partners at r ≥ 0.5, and the chance level from the same data with each column permuted independently once.

```python
# Step 0: premise check on the 80- and the 79-sample data. D from the real labels; enrichment of
# D among the correlation partners of D proteins, for two within-group correlation matrices
seed_fdr = 0.05


def top_partners(corr, m=3):
    """Each protein's m most positively correlated other proteins (ranked as in rule A)."""
    c = np.where(np.isfinite(corr), corr, -np.inf)
    np.fill_diagonal(c, -np.inf)
    ranked = np.argsort(-c, axis=1)[:, :m]
    return ranked, np.take_along_axis(c, ranked, axis=1) > 0   # partners, and whether r > 0


def top_share(ranked, positive, members, of):
    """Share of members among the positive top partners of the proteins in `of`; pair count."""
    partners = ranked[of][positive[of]]
    return members[partners].mean(), len(partners)


def linked_share(strong, members):
    """P(protein in members | strong link to some OTHER member); strong has a False diagonal."""
    linked = strong[:, members].any(axis=1)
    return members[linked].mean(), int(linked.sum()), members[~linked].mean()


def premise_metrics(data, in_D, seeds, n_perm=1000):
    """Enrichment, permutation baseline and correlation noise for one (samples x proteins) matrix."""
    p, n_D = data.shape[1], int(in_D.sum())
    corr = np.corrcoef(data, rowvar=False)
    ranked, positive = top_partners(corr)
    strong = corr >= 0.5
    np.fill_diagonal(strong, False)

    # Enrichment, observed
    share_D, n_pairs_D = top_share(ranked, positive, in_D, in_D)
    share_nonD, n_pairs_nonD = top_share(ranked, positive, in_D, ~in_D)
    link_D, n_linked, unlinked_D = linked_share(strong, in_D)
    any_link = strong.any(axis=1)
    rows = [("n_samples_matrix", data.shape[0], np.nan),
            ("top3_share_D_among_partners_of_D", share_D, n_pairs_D),
            ("top3_share_D_among_partners_of_nonD", share_nonD, n_pairs_nonD),
            ("r05_P_D_given_link_to_other_D", link_D, n_linked),
            ("r05_P_D_given_no_link_to_D", unlinked_D, p - n_linked),
            ("r05_P_D_given_any_link", in_D[any_link].mean(), int(any_link.sum()))]

    # Permutation baseline: random sets of the same size as D
    rng = np.random.default_rng([4200, *seeds])
    perm_top, perm_link = np.empty(n_perm), np.empty(n_perm)
    for b in range(n_perm):
        members = np.zeros(p, dtype=bool)
        members[rng.choice(p, n_D, replace=False)] = True
        perm_top[b] = top_share(ranked, positive, members, members)[0]
        perm_link[b] = linked_share(strong, members)[0]
    rows += [("top3_perm_mean", perm_top.mean(), n_perm),
             ("top3_perm_q95", np.quantile(perm_top, 0.95), n_perm),
             ("top3_perm_p", (1 + (perm_top >= share_D).sum()) / (1 + n_perm), n_perm),
             ("r05_perm_mean", np.nanmean(perm_link), n_perm),
             ("r05_perm_q95", np.nanquantile(perm_link, 0.95), n_perm),
             ("r05_perm_p", (1 + (perm_link >= link_D).sum()) / (1 + n_perm), n_perm)]

    # Correlation noise
    off = corr[np.triu_indices(p, 1)]
    c = corr.copy()
    np.fill_diagonal(c, np.nan)
    max_r = np.nanmax(c, axis=1)
    n_partners = strong.sum(axis=1)
    rows += [(f"offdiag_r_q{q:g}", np.quantile(off, q), len(off))
             for q in [0.01, 0.05, 0.25, 0.5, 0.75, 0.95, 0.99]]
    rows += [("offdiag_r_mean", off.mean(), len(off)),
             ("n_pairs_r05", int((off >= 0.5).sum()), len(off))]
    rows += [(f"max_r_per_protein_q{q:g}", np.quantile(max_r, q), p) for q in [0.05, 0.25, 0.5, 0.75, 0.95]]
    rows += [(f"n_partners_r05_q{q:g}", np.quantile(n_partners, q), p) for q in [0.25, 0.5, 0.75, 0.95]]
    rows += [("n_partners_r05_max", n_partners.max(), p),
             ("share_proteins_0_partners_r05", (n_partners == 0).mean(), p)]

    # Chance level: each column permuted independently once (no correlation left)
    rng = np.random.default_rng([4201, *seeds])
    chance = np.corrcoef(np.column_stack([rng.permutation(col) for col in data.T]), rowvar=False)
    np.fill_diagonal(chance, np.nan)
    chance_max_r, chance_max_abs = np.nanmax(chance, axis=1), np.nanmax(np.abs(chance), axis=1)
    rows += [("chance_max_r_per_protein_q0.5", np.median(chance_max_r), p),
             ("chance_max_r_per_protein_q0.95", np.quantile(chance_max_r, 0.95), p),
             ("chance_max_abs_r_per_protein_q0.5", np.median(chance_max_abs), p),
             ("chance_max_abs_r_per_protein_q0.95", np.quantile(chance_max_abs, 0.95), p),
             ("chance_max_abs_r_overall", np.nanmax(np.abs(chance)), p),
             ("chance_share_proteins_with_partner_r05", (chance >= 0.5).any(axis=1).mean(), p)]
    return rows


premise_rows = []
for samples, df in {80: dia_all_imp, 79: seed_data}.items():
    is_cancer = (df["GROUP"] == "Cancer").to_numpy()
    X = df[seed_prots].to_numpy(dtype=float)
    in_D = _bh_adjust(welch_pvalues(X[~is_cancer], X[is_cancer])) < seed_fdr
    print(f"{samples} samples: |D| = {in_D.sum()} of {seed_p} proteins (BH {seed_fdr}); "
          f"base rate {in_D.mean():.3f}")
    premise_rows += [(samples, "labels", "n_samples", len(df), np.nan),
                     (samples, "labels", "n_D", int(in_D.sum()), np.nan),
                     (samples, "labels", "base_rate", in_D.mean(), np.nan)]
    matrices = {"centered_all": center_by_group(df, "GROUP")[seed_prots].to_numpy(),
                "normal_only": X[~is_cancer]}
    for mi, (matrix, data) in enumerate(matrices.items()):
        premise_rows += [(samples, matrix, metric, value, n)
                         for metric, value, n in premise_metrics(data, in_D, (samples, mi))]

seed_premise = pd.DataFrame(premise_rows, columns=["samples", "matrix", "metric", "value", "n"])
seed_premise.to_csv("../data/results/PDAC_DIA/sample_size_seed_premise.csv", index=False)
with pd.option_context("display.max_rows", 200):
    display(seed_premise.pivot_table(index="metric", columns=["matrix", "samples"], values="value",
                                     sort=False).round(3))

# 4.2 null (79 samples): the partners and the Copula fits come from it
seed_null = center_by_group(seed_data, "GROUP")[seed_prots]
seed_corr = seed_null.corr().to_numpy()
```

### Changed sets and experiments

Devon, 2026-10-16: the correlation-removed Copula uses a fixed 5% of proteins (20 of 397) changed at random, the independence planner's assumption; the benchmark and Copula use the seed sets.

`seed_partner_links` re-derives which seed added each partner (same selection as `seed_changed_sets`, checked against its output) to record each partner's correlation to its seed (rule B: the strongest seed that selected it). Seeds: seed sets `[4206, rule, N]`, random sets of copula_independent `[4208, N]`, rows `[4207, rule, log2FC, N, pool]` (indices into the grids). `sample_size_seed_sets.csv` holds both: the seed sets under rule A and B, and the random sets under rule = "random" (seeds empty, partner columns NaN); copula_independent rows of either rule use the random set with the same N and experiment index.

```python
# Grid, changed sets and the experiment loop
seed_rules = {"A": dict(top_m=3), "B": dict(min_corr=0.5, max_partners=15)}
seed_n_seeds = 5
seed_log2fc_grid = [0.58, 1.0]
seed_n_grid = [3, 5, 8, 10, 15, 20, 25, 30, 39]   # 2N <= 79 benchmark rows
seed_n_experiments = 500
seed_methods = list(seed_pools)   # benchmark, copula, copula_independent
seed_random_k = 20   # copula_independent: fixed 5% of 397 changed at random (Devon, 2026-10-16)
seed_experiment_cols = ["rule", "method", "log2fc", "n_per_group", "experiment", "n_changed",
                        "tp", "fp", "discoveries", "power", "fdp"]

# Partner ranking exactly as in seed_changed_sets
seed_corr_ranked = np.where(np.isfinite(seed_corr), seed_corr, -np.inf)
np.fill_diagonal(seed_corr_ranked, -np.inf)
seed_ranked = np.argsort(-seed_corr_ranked, axis=1)


def seed_partner_links(seeds, top_m=None, min_corr=None, max_partners=None):
    """The changed set of these seeds, and each non-seed member's correlation to its seed."""
    seed_set = set(seeds.tolist())
    changed, link = set(seed_set), {}
    for s in seeds:
        partners = [j for j in seed_ranked[s] if seed_corr_ranked[s, j] > 0]
        if top_m is not None:
            partners = [j for j in partners if j not in changed][:top_m]
        else:
            partners = [j for j in partners if seed_corr_ranked[s, j] >= min_corr][:max_partners]
        changed.update(partners)
        for j in partners:
            if j not in seed_set:
                link[j] = max(link.get(j, -np.inf), seed_corr_ranked[s, j])
    return changed, np.array(list(link.values()))


def draw_seed_sets(rule, n):
    """The 500 changed sets of one (rule, N) and a table of their seeds and partner correlations."""
    ri, ni = list(seed_rules).index(rule), seed_n_grid.index(n)
    sets, seeds = seed_changed_sets(seed_corr, seed_n_experiments, seed_n_seeds,
                                    np.random.default_rng([4206, ri, ni]), **seed_rules[rule])
    rows = []
    for e, (members, s) in enumerate(zip(sets, seeds)):
        changed, r = seed_partner_links(s, **seed_rules[rule])
        assert changed == set(members.tolist()), (rule, n, e)
        rows.append(dict(rule=rule, n_per_group=n, experiment=e, n_changed=len(members),
                         n_partners=len(r), seeds=";".join(seed_prots[i] for i in s),
                         partner_r_mean=r.mean() if len(r) else np.nan,
                         partner_r_min=r.min() if len(r) else np.nan,
                         changed_proteins=";".join(seed_prots[i] for i in members)))
    return sets, pd.DataFrame(rows)


def draw_random_sets(n):
    """The 500 random sets of copula_independent for one N (20 proteins, uniform, no replacement)."""
    rng = np.random.default_rng([4208, seed_n_grid.index(n)])
    sets = [np.sort(rng.choice(seed_p, seed_random_k, replace=False))
            for _ in range(seed_n_experiments)]
    table = pd.DataFrame(dict(rule="random", n_per_group=n, experiment=np.arange(len(sets)),
                              n_changed=seed_random_k, n_partners=np.nan, seeds="",
                              partner_r_mean=np.nan, partner_r_min=np.nan,
                              changed_proteins=[";".join(seed_prots[i] for i in s) for s in sets]))
    return sets, table


def run_seed_grid(n_grid, n_experiments):
    """Experiments for every (rule, N, log2FC, pool); returns experiments, summary and sets."""
    experiments, summary, set_tables, random_tables = [], [], [], []
    random_sets = {}
    for n in n_grid:
        sets, table = draw_random_sets(n)
        random_sets[n] = sets[:n_experiments]
        random_tables.append(table.iloc[:n_experiments])
    for ri, rule in enumerate(seed_rules):
        for n in n_grid:
            ni = seed_n_grid.index(n)
            sets, set_table = draw_seed_sets(rule, n)
            sets = sets[:n_experiments]
            set_tables.append(set_table.iloc[:n_experiments])
            for fi, log2fc in enumerate(seed_log2fc_grid):
                for mi, method in enumerate(seed_methods):
                    # copula_independent: random sets; benchmark and copula: seed sets
                    method_sets = random_sets[n] if method == "copula_independent" else sets
                    rng = np.random.default_rng([4207, ri, fi, ni, mi])
                    exps, _ = run_experiments(seed_pools[method], n, None, log2fc, n_experiments,
                                              rng, fdr=seed_fdr, changed_sets=method_sets)
                    design = dict(rule=rule, method=method, log2fc=log2fc, n_per_group=n)
                    experiments.append(exps.assign(experiment=np.arange(n_experiments), **design))
                    summary.append({**design, "mean_n_changed": exps["n_changed"].mean(),
                                    "mean_tp": exps["tp"].mean(), "mean_fp": exps["fp"].mean(),
                                    **summarize_experiments(exps).to_dict()})
    return (pd.concat(experiments, ignore_index=True)[seed_experiment_cols],
            pd.DataFrame(summary), pd.concat(set_tables + random_tables, ignore_index=True))
```

```python
# Smoke test (optional, seconds): N = 10, the first 20 changed sets per rule. Nothing is saved.
t0 = time.time()
seed_smoke, seed_smoke_summary, _ = run_seed_grid([10], 20)
print(f"smoke test: {len(seed_smoke)} experiments in {time.time() - t0:.1f} s")
seed_smoke_summary.round(3)
```

```python
# Full run: 500 experiments per (rule, log2FC, N, pool); saves experiments, summary and sets
t0 = time.time()
seed_experiments, seed_summary, seed_sets = run_seed_grid(seed_n_grid, seed_n_experiments)
print(f"full run: {len(seed_experiments)} experiments in {time.time() - t0:.0f} s")
seed_experiments.to_csv("../data/results/PDAC_DIA/sample_size_seed_experiments.csv", index=False)
seed_summary.to_csv("../data/results/PDAC_DIA/sample_size_seed_summary.csv", index=False)
seed_sets.to_csv("../data/results/PDAC_DIA/sample_size_seed_sets.csv", index=False)
print("experiments:", seed_experiments.shape, "| summary:", seed_summary.shape,
      "| sets:", seed_sets.shape)
```
